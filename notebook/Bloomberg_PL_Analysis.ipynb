```python
def clean_value(value):
    if pd.isna(value):
        return np.nan
    
    value = str(value).strip()
    
    if value in ["NM", "N/A", "#N/A N/A", "—", "-", ""]:
        return np.nan
    
    if value.endswith("%"):
        return float(value.replace("%", "").replace(",", "").strip()) / 100
    
    if value.startswith("(") and value.endswith(")"):
        value = "-" + value[1:-1]
    
    value = value.replace(",", "").strip()
    
    return float(value)
```

```python
test_values = [
    "11,663.00",
    "-9,500.00",
    "14.80%",
    "(392.00)",
    "NM",
    "—",
    "#N/A N/A"
]

for value in test_values:
    print(value, "→", clean_value(value))
```
